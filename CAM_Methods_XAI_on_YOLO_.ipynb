```python
import os
import cv2
import numpy as np
import pandas as pd
from PIL import Image
from fpdf import FPDF
from YOLOv8_Explainer import yolov8_heatmap

# Image and model config
img_folder = "Add path of validation images folder"  # Update if needed
jpg_files = [os.path.join(img_folder, f) for f in os.listdir(img_folder) if f.endswith('.jpg')]
model_path = "Add path to the trained YOLO model"

# CAM methods to test
cam_methods = ["GradCAM", "HiResCAM", "GradCAMPlusPlus", "XGradCAM", "LayerCAM", "EigenGradCAM", "EigenCAM"]

# Output folders
output_base = "Add path you need to store output in"
output_img_dir = os.path.join(output_base, "images")
os.makedirs(output_img_dir, exist_ok=True)

results = []
```

```python
def pointing_game(cam_map, bboxes):
    """
    Checks if the CAM peak lies inside any bounding box.
    Returns 1 if yes (hit), 0 otherwise.
    """
    h, w = cam_map.shape
    max_pos = np.unravel_index(np.argmax(cam_map), cam_map.shape)
    y, x = max_pos  # row, col

    for box in bboxes:
        x1, y1, x2, y2 = map(int, box)
        if x1 <= x <= x2 and y1 <= y <= y2:
            return 1  # Hit
    return 0  # Miss


def iou_score(cam_map, bboxes, threshold=0.5):
    """
    Computes IoU between binarized CAM and bounding boxes.
    Returns average IoU over all boxes.
    """
    cam_bin = cam_map > threshold
    iou_scores = []

    for box in bboxes:
        x1, y1, x2, y2 = map(int, box)
        x1 = max(0, x1)
        x2 = min(cam_map.shape[1], x2)
        y1 = max(0, y1)
        y2 = min(cam_map.shape[0], y2)

        bbox_mask = np.zeros_like(cam_map, dtype=bool)
        bbox_mask[y1:y2, x1:x2] = True

        intersection = np.logical_and(cam_bin, bbox_mask).sum()
        union = np.logical_or(cam_bin, bbox_mask).sum()
        iou = intersection / (union + 1e-8)
        iou_scores.append(iou)

    return np.mean(iou_scores) if iou_scores else 0.0


def compute_focus_score(cam_map, bboxes):
    """
    Measures how much of the CAM energy lies inside the bounding boxes.
    """
    h, w = cam_map.shape
    focus_scores = []

    for box in bboxes:
        x1, y1, x2, y2 = map(int, box)
        x1 = max(0, min(w - 1, x1))
        x2 = max(0, min(w - 1, x2))
        y1 = max(0, min(h - 1, y1))
        y2 = max(0, min(h - 1, y2))

        cam_inside = cam_map[y1:y2, x1:x2]
        focus = np.sum(cam_inside) / (np.sum(cam_map) + 1e-8)
        focus_scores.append(focus)

    return np.mean(focus_scores) if focus_scores else 0.0


def compute_sharpness(cam_map):
    cam_uint8 = np.uint8(255 * cam_map) if cam_map.max() <= 1 else cam_map.astype(np.uint8)
    laplacian = cv2.Laplacian(cam_uint8, cv2.CV_64F)
    return laplacian.var()
```

```python
from ultralytics import YOLO

results = []

for method in cam_methods:
    print(f"🔍 Running CAM Method: {method}")

    # Create a folder for each CAM method
    method_output_dir = os.path.join(output_img_dir, method)
    os.makedirs(method_output_dir, exist_ok=True)

    # Load model with specific CAM
    model = yolov8_heatmap(
        weight=model_path,
        conf_threshold=0.4,
        method=method,
        layer=[10, 12, 14, 16, 18, -3],
        ratio=0.02,
        show_box=False,
        renormalize=False,
    )

    for img_path in jpg_files:
        img_name = os.path.basename(img_path)
        result = model(img_path=img_path)

        if not result or result[0] is None:
            print(f"⚠️ No output for {img_name} using {method}")
            continue

        for i, img_np in enumerate(result):
            output_file = os.path.join(method_output_dir, img_name)
            img_np.save(output_file)

            # Convert image to array and grayscale
            img_array = np.array(img_np)
            gray = cv2.cvtColor(img_array, cv2.COLOR_RGB2GRAY)
            sharpness = compute_sharpness(gray)

            # Get CAM heatmap for metrics
            cam_map = gray / 255.0  # Normalize CAM for metric use

            yolo_model = YOLO(model_path)  # load the original model separately

            # Later inside your loop
            preds = yolo_model(img_path)[0]
            # preds = model(img_path)[0]
            bboxes = preds.boxes.xyxy.cpu().numpy()

            # Calculate metrics
            pg_hit = pointing_game(cam_map, bboxes)
            iou = iou_score(cam_map, bboxes)
            focus = compute_focus_score(cam_map, bboxes)

            results.append({
                "Image": img_name,
                "Method": method,
                "Focus Score": round(focus, 4),
                "PG_Hit": round(pg_hit, 2),
                "IoU": round(iou, 4),
                "Sharpness": round(sharpness, 2),
                "Output Path": output_file
            })
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Convert results to DataFrame
df = pd.DataFrame(results)

# Preview all metrics
print("📊 Sample of all metric results:")
display(df.head(10))  # Show top 10 rows

# Group by method and show average for each metric
avg_metrics = df.groupby("Method").mean(numeric_only=True)[["Focus Score", "PG_Hit", "IoU", "Sharpness"]]
print("📈 Average Metrics Per Method:")
display(avg_metrics)

# Optional plot
avg_metrics.plot(kind="bar", figsize=(10, 6), title="Average CAM Metrics per Method")
plt.ylabel("Score")
plt.xticks(rotation=45)
plt.grid(axis="y")
plt.tight_layout()
plt.show()
```
